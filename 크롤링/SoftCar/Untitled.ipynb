```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time

# 크롬 드라이버 설정
chrome_options = Options()
chrome_options.add_argument("--headless")  # 창을 띄우지 않고 실행
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")

# 크롬 드라이버 경로 설정
chrome_driver_path = '/path/to/chromedriver'  # 크롬 드라이버 경로 설정

# 웹 드라이버 서비스 시작
service = Service(chrome_driver_path)
driver = webdriver.Chrome(service=service, options=chrome_options)

# 그린카 웹사이트 열기
driver.get('https://www.greencar.co.kr/main/')

# 로그인 (필요 시)
username = 'your_username'
password = 'YOUR_CREDENTIAL_HERE'

# 로그인 페이지로 이동
login_button = driver.find_element(By.XPATH, '//*[@id="login_btn"]')
login_button.click()

# 로그인 정보 입력
driver.find_element(By.NAME, 'id').send_keys(username)
driver.find_element(By.NAME, 'password').send_keys(password)
driver.find_element(By.NAME, 'password').send_keys(Keys.RETURN)

time.sleep(3)  # 로그인 처리 시간을 기다림

# 차고지 위치 페이지로 이동
driver.get('https://www.greencar.co.kr/garage/')  # 실제 차고지 위치 페이지 URL로 수정 필요

# 차고지 위치 정보 크롤링
time.sleep(3)  # 페이지 로딩 시간을 기다림

# 차고지 위치 정보를 담고 있는 요소 찾기
garage_elements = driver.find_elements(By.CLASS_NAME, 'garage_class')  # 실제 클래스명으로 수정 필요

garages = []
for element in garage_elements:
    location = element.text
    garages.append(location)

# 크롤링 결과 출력
for garage in garages:
    print(garage)

# 웹 드라이버 종료
driver.quit()
```
