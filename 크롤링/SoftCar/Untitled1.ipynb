```python
import requests
from bs4 import BeautifulSoup

def crawl_green_car_locations():
    url = 'https://www.greencar.co.kr/'
    
    try:
        response = requests.get(url)
        response.raise_for_status()  # HTTP 오류 발생 시 예외 발생
    except requests.exceptions.RequestException as e:
        print(f"HTTP 요청 오류: {e}")
        return
    
    html = response.text
    soup = BeautifulSoup(html, 'html.parser')
    
    # 차고지 위치를 포함하는 요소들 찾기
    locations = soup.find_all('div', class_='garage-box')
    
    if not locations:
        print("차고지 위치를 찾을 수 없습니다.")
        return
    
    # 위치 정보 출력
    for location in locations:
        name = location.find('h3').text.strip()
        address = location.find('p', class_='location-address').text.strip()
        print(f'차고지 이름: {name}')
        print(f'주소: {address}')
        print('---')

crawl_green_car_locations()
```
