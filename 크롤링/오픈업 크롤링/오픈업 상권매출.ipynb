```python
from selenium.webdriver.common.by import By
import time
import pandas as pd

from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from selenium.webdriver.chrome.options import Options

# 웹드라이버 초기화
chrome_options = Options()
chrome_options.add_argument("--start-fullscreen")  # 브라우저를 전체화면 모드로 시작
driver = webdriver.Chrome(service=ChromeService(), options=chrome_options)
driver.get("https://www.openub.com/")

# 버튼 클릭
button = driver.find_element(By.CSS_SELECTOR, "button.close.p-8-0.bg-transparent.font-14.text-gray-700.text-center.font-bold.cursor-pointer.hover\\:bg-black-0_03")
button.click()

button2 = driver.find_element(By.CSS_SELECTOR, "button.close.pos-rel.w-100p.p-10-8.bg-white.brdrBR-4.font-14.text-gray-600.text-center.cursor-pointer.hover\\:before\\:bg-black-0_03")
button2.click()

zoom_in_button = driver.find_element(By.CSS_SELECTOR, "button.right-icon-zoomin.icon-zoomin")
zoom_in_button.click()

for _ in range(3):
    zoom_in_button.click()
    # 각 클릭 사이에 잠시 대기 (필요에 따라 조정 가능)
    time.sleep(1)
    
# 아이콘을 찾기 위한 CSS 선택자
icons = driver.find_elements(By.CSS_SELECTOR, "div.z-1.hover\\:z-4")

# 아이콘 클릭 후 매출액과 주소 정보 추출 함수
def extract_info(icon):
    icon.click()
    time.sleep(1)  # 페이지가 반응할 시간을 제공
    
    try:
        # 매출액 정보 추출
        sales_info = driver.find_element(By.CSS_SELECTOR, "p#tab-bd-month-sales")
        sales_text = sales_info.text
    except:
        sales_text = "매출액 정보 없음"
    
    try:
        # 주소 정보 추출
        address_info = driver.find_element(By.CSS_SELECTOR, "p.address.font-14.text-gray-700.text-ellipsis")
        address_text = address_info.text
    except:
        address_text = "주소 정보 없음"
    
    return sales_text, address_text

# 결과를 저장할 리스트
results = []

# 모든 아이콘에 대해 매출액과 주소 정보 추출
for index, icon in enumerate(icons):
    try:
        sales_text, address_text = extract_info(icon)
        results.append({
            "아이콘 번호": index + 1,
            "매출액": sales_text,
            "주소": address_text
        })
    except Exception as e:
        print(f"아이콘 {index + 1}에서 정보 추출 실패: {e}")

# 데이터프레임 생성
df = pd.DataFrame(results)

# 데이터프레임을 CSV 파일로 저장
df.to_csv('sales_and_addresses.csv', index=False, encoding='utf-8-sig')

# 브라우저 종료
driver.quit()
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd

from selenium import webdriver
from selenium.webdriver.chrome.service import Service as ChromeService
from selenium.webdriver.chrome.options import Options

# 웹드라이버 초기화
chrome_options = Options()
chrome_options.add_argument("--start-fullscreen")  # 브라우저를 전체화면 모드로 시작
driver = webdriver.Chrome(service=ChromeService(), options=chrome_options)
driver.get("https://www.openub.com/")

# 버튼 클릭
try:
    button = driver.find_element(By.CSS_SELECTOR, "button.close.p-8-0.bg-transparent.font-14.text-gray-700.text-center.font-bold.cursor-pointer.hover\\:bg-black-0_03")
    button.click()
except Exception as e:
    print(f"첫 번째 버튼 클릭 실패: {e}")

try:
    button2 = driver.find_element(By.CSS_SELECTOR, "button.close.pos-rel.w-100p.p-10-8.bg-white.brdrBR-4.font-14.text-gray-600.text-center.cursor-pointer.hover\\:before\\:bg-black-0_03")
    button2.click()
except Exception as e:
    print(f"두 번째 버튼 클릭 실패: {e}")

# 줌 인 버튼 클릭
try:
    zoom_in_button = driver.find_element(By.CSS_SELECTOR, "button.right-icon-zoomin.icon-zoomin")
    for _ in range(3):
        zoom_in_button.click()
        time.sleep(1)
except Exception as e:
    print(f"줌 인 버튼 클릭 실패: {e}")

# 아이콘 클릭 후 매출액과 주소 정보 추출 함수
def extract_info(icon):
    icon.click()
    time.sleep(2)  # 페이지가 반응할 시간을 제공
    
    try:
        sales_info = driver.find_element(By.CSS_SELECTOR, "p#tab-bd-month-sales")
        sales_text = sales_info.text
    except:
        sales_text = "매출액 정보 없음"
    
    try:
        address_info = driver.find_element(By.CSS_SELECTOR, "p.address.font-14.text-gray-700.text-ellipsis")
        address_text = address_info.text
    except:
        address_text = "주소 정보 없음"
    
    return sales_text, address_text

# 지도 이동 및 아이콘 찾기 함수
def move_map_and_crawl():
    results = []
    directions = [Keys.ARROW_RIGHT, Keys.ARROW_DOWN, Keys.ARROW_LEFT, Keys.ARROW_UP]
    num_moves = 0

    while num_moves < 10:  # 최대 이동 횟수 설정
        icons = driver.find_elements(By.CSS_SELECTOR, "div.z-1.hover\\:z-4")
        for index, icon in enumerate(icons):
            try:
                sales_text, address_text = extract_info(icon)
                results.append({
                    "아이콘 번호": len(results) + 1,
                    "매출액": sales_text,
                    "주소": address_text
                })
            except Exception as e:
                print(f"아이콘 {index + 1}에서 정보 추출 실패: {e}")

        # 방향키를 사용하여 지도 이동
        for direction in directions:
            action = ActionChains(driver)
            action.send_keys(direction).perform()
            time.sleep(2)  # 지도 이동 후 아이콘 로딩 대기

        num_moves += 1
    
    return results

# 데이터 추출
results = move_map_and_crawl()

# 데이터프레임 생성
df = pd.DataFrame(results)

# 데이터프레임을 CSV 파일로 저장
df.to_csv('sales_and_addresses2.csv', index=False, encoding='utf-8-sig')

# 브라우저 종료
driver.quit()
```

```python
import pandas as pd
import numpy as np
import re

# Load the data
df = pd.read_csv("sales_and_addresses_final.csv")

def convert_sales_to_number(sales_str):
    # Remove '약 ' and '원'
    sales_str = sales_str.replace('약 ', '').replace('원', '')
    
    # Remove commas and spaces
    sales_str = sales_str.replace(',', '').replace(' ', '')
    
    # Initialize variables for the final number
    total = 0

    # Handle cases where '억' is present
    if '억' in sales_str:
        parts = sales_str.split('억')
        if parts[0]:
            try:
                total += float(parts[0]) * 100_000_000
            except ValueError:
                return np.nan
        if len(parts) > 1 and '만' in parts[1]:
            # Process the '만' part if present
            sub_parts = parts[1].split('만')
            if sub_parts[0]:
                try:
                    total += float(sub_parts[0]) * 10_000
                except ValueError:
                    return np.nan
    elif '만' in sales_str:
        # Handle cases where only '만' is present
        parts = sales_str.split('만')
        if parts[0]:
            try:
                total += float(parts[0]) * 10_000
            except ValueError:
                return np.nan
    else:
        # Handle cases where neither '억' nor '만' is present
        try:
            total = float(sales_str)
        except ValueError:
            return np.nan
    
    return total

# Apply the conversion function to the '매출액' column
df['매출액_숫자'] = df['매출액'].apply(convert_sales_to_number)

# Save the cleaned data
df.to_csv("sales_and_addresses_final_cleaned.csv", index=False, encoding = "utf-8-sig")

# Print a sample of the cleaned data
print(df[['매출액', '매출액_숫자']].head())
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderServiceError
import time

# CSV 파일 경로
input_file_path = r'C:\Users\kfq\Downloads\1ST_Final_Project\크롤링\오픈업 크롤링\sales_and_addresses_final_cleaned.csv'
output_file_path = r'C:\Users\kfq\Downloads\1ST_Final_Project\크롤링\오픈업 크롤링\sales_and_addresses_final_cleaned_geocode.csv'

# CSV 파일 읽기
try:
    df = pd.read_csv(input_file_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(input_file_path, encoding='cp949')

# 지오코더 초기화
geolocator = Nominatim(user_agent="socar_geocoder", timeout=10)  # timeout 값을 10초로 설정

# 지오코딩 함수 정의
def geocode_address(address, retries=5):
    for attempt in range(retries):
        try:
            location = geolocator.geocode(address)
            if location:
                return location.latitude, location.longitude
        except GeocoderTimedOut:
            print(f"Geocoding timed out on attempt {attempt + 1}. Retrying...")
            time.sleep(2)  # 2초 대기 후 재시도
        except GeocoderServiceError as e:
            print(f"Geocoding service error: {e}")
        except Exception as e:
            print(f"Unexpected error: {e}")
    return None, None  # 재시도 후에도 실패하면 None 반환

# 위도와 경도 컬럼 추가
latitudes = []
longitudes = []

for index, row in df.iterrows():
    lat, lon = None, None
    if pd.notna(row['주소']):
        lat, lon = geocode_address(row['주소'])
    
    latitudes.append(lat)
    longitudes.append(lon)

df['위도'] = latitudes
df['경도'] = longitudes

# 위도와 경도만 포함된 새로운 데이터프레임 생성
df_geocoded = df[['위도', '경도']]

# 결과를 새로운 CSV 파일로 저장 (utf-8 인코딩)
df_geocoded.to_csv(output_file_path, index=False, encoding='utf-8-sig')

print("지오코딩이 완료되었습니다. 결과가 저장되었습니다:", output_file_path)
```

```python
# 인천광역시 행정동별 매출액.csv

import pandas as pd
import numpy as np
import re

# Load the data
df = pd.read_csv("대전광역시 상권매출 (오픈업).csv")

def convert_sales_to_number(sales_str):
    # Remove '약 ' and '원'
    sales_str = sales_str.replace('약 ', '').replace('원', '')
    
    # Remove commas and spaces
    sales_str = sales_str.replace(',', '').replace(' ', '')
    
    # Initialize variables for the final number
    total = 0

    # Handle cases where '억' is present
    if '억' in sales_str:
        parts = sales_str.split('억')
        if parts[0]:
            try:
                total += float(parts[0]) * 100_000_000
            except ValueError:
                return np.nan
        if len(parts) > 1 and '만' in parts[1]:
            # Process the '만' part if present
            sub_parts = parts[1].split('만')
            if sub_parts[0]:
                try:
                    total += float(sub_parts[0]) * 10_000
                except ValueError:
                    return np.nan
    elif '만' in sales_str:
        # Handle cases where only '만' is present
        parts = sales_str.split('만')
        if parts[0]:
            try:
                total += float(parts[0]) * 10_000
            except ValueError:
                return np.nan
    else:
        # Handle cases where neither '억' nor '만' is present
        try:
            total = float(sales_str)
        except ValueError:
            return np.nan
    
    return total

# Apply the conversion function to the '매출액' column
df['매출액_숫자'] = df['매출액'].apply(convert_sales_to_number)

# Save the cleaned data
df.to_csv("인천광역시 행정동별 매출액_int.csv", index=False, encoding = "utf-8-sig")

# Print a sample of the cleaned data
print(df[['매출액', '매출액_숫자']].head())
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderServiceError
import time

# CSV 파일 경로
input_file_path = r'C:\Users\kfq\Downloads\1ST_Final_Project\크롤링\오픈업 크롤링\대전광역시 상권매출 (오픈업).csv'
output_file_path = r'C:\Users\kfq\Downloads\1ST_Final_Project\크롤링\오픈업 크롤링\대전광역시 상권매출 (오픈업)_geocode.csv'

# CSV 파일 읽기
try:
    df = pd.read_csv(input_file_path, encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv(input_file_path, encoding='cp949')

# 지오코더 초기화
geolocator = Nominatim(user_agent="socar_geocoder", timeout=10)  # timeout 값을 10초로 설정

# 지오코딩 함수 정의
def geocode_address(address, retries=5):
    for attempt in range(retries):
        try:
            location = geolocator.geocode(address)
            if location:
                return location.latitude, location.longitude
        except GeocoderTimedOut:
            print(f"Geocoding timed out on attempt {attempt + 1}. Retrying...")
            time.sleep(2)  # 2초 대기 후 재시도
        except GeocoderServiceError as e:
            print(f"Geocoding service error: {e}")
        except Exception as e:
            print(f"Unexpected error: {e}")
    return None, None  # 재시도 후에도 실패하면 None 반환

# 위도와 경도 컬럼 추가
latitudes = []
longitudes = []

for index, row in df.iterrows():
    lat, lon = None, None
    if pd.notna(row['주소']):
        lat, lon = geocode_address(row['주소'])
    
    latitudes.append(lat)
    longitudes.append(lon)

df['위도'] = latitudes
df['경도'] = longitudes

# 위도와 경도만 포함된 새로운 데이터프레임 생성
df_geocoded = df[['위도', '경도']]

# 결과를 새로운 CSV 파일로 저장 (utf-8 인코딩)
df_geocoded.to_csv(output_file_path, index=False, encoding='utf-8-sig')

print("지오코딩이 완료되었습니다. 결과가 저장되었습니다:", output_file_path)
```
