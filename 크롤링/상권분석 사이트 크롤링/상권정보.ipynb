```python
from selenium import webdriver as wb
from selenium.webdriver.common.action_chains import ActionChains
import time

# 웹드라이버 초기화
driver = wb.Chrome()

# 웹페이지 열기
driver.get("https://sg.sbiz.or.kr/godo/index.sg")

# 페이지가 로드될 시간을 기다림
time.sleep(5)  # 페이지 로드 시간에 따라 조정할 수 있습니다.

# 첫 번째 팝업 닫기 (화면의 특정 위치 클릭)
x_coordinate_1 = 100
y_coordinate_1 = 100

actions = ActionChains(driver)
actions.move_by_offset(x_coordinate_1, y_coordinate_1).click().perform()

# 첫 번째 클릭 후 대기 (팝업이 사라지기를 기다림)
time.sleep(2)  # 대기 시간 조정 가능

# 두 번째 팝업 닫기 (화면의 특정 위치 클릭)
x_coordinate_2 = 200
y_coordinate_2 = 200

actions.move_by_offset(x_coordinate_2, y_coordinate_2).click().perform()

try:
    # 첫 번째 요소 클릭 (리스트 아이템)
    list_item = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "li[data-icon='d02']"))
    )
    list_item.click()
    
    # 두 번째 요소 클릭 (텍스트 입력 버튼)
    text_button = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.ID, "upjongSelection"))
    )
    text_button.click()

    # 세 번째 요소 클릭 (리스트 아이템 클래스 'hit')
    hit_item = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "li.hit"))
    )
    hit_item.click()
    
    # 네 번째 요소 클릭 (확인 버튼 - yes)
    # 스크롤하여 요소가 화면에 보이도록 하기
    confirm_button = WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "div.foot a#checkTypeConfirm[data-name='yes']"))
    )
    driver.execute_script("arguments[0].scrollIntoView(true);", confirm_button)
    
    # 잠시 대기 후 클릭 시도
    time.sleep(2)  # 요소가 클릭 가능할 때까지 대기

    # JavaScript를 사용하여 클릭하기 (다른 요소가 방해할 때 유용)
    driver.execute_script("arguments[0].click();", confirm_button)

except Exception as e:
    print(f"오류 발생: {e}")
# driver.quit()
```
