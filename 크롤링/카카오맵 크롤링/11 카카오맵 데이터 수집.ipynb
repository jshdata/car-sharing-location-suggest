```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup as bs

# webdriver_manager를 사용하여 Chrome 드라이버 설정
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# 카카오 지도 웹사이트 열기
driver.get("https://map.kakao.com/")

# 요소가 나타날 때까지 기다린 후 클릭
time.sleep(2)
img = driver.find_element(By.CSS_SELECTOR, "div.inner_coach_layer")
img.click()

# 검색 수행
search = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.query")
search.send_keys("쏘카")
search_btn = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.submit")
search_btn.click()

# 검색 결과가 로드될 시간을 줌
time.sleep(2)

# BeautifulSoup을 사용하여 페이지 소스 파싱
soup = bs(driver.page_source, "lxml")

# 광고 제거
ads = soup.select("span.ico_ad")
for ad in ads:
    ad.extract()

# "장소 더보기" 버튼 클릭
btn_more = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.place\\.more")
btn_more.click()

# 새로운 결과가 로드될 시간을 줌
time.sleep(2)

try:
    while True:
        for page_num in range(1, 6):
            # 페이지 버튼 클릭 후 데이터 스크랩
            page_button = driver.find_element(By.CSS_SELECTOR, f"#info\\.search\\.page\\.no{page_num}")
            page_button.click()
            time.sleep(2)
            
            soup = bs(driver.page_source, "lxml")
            
            # 광고 제거
            ads = soup.select("span.ico_ad")
            for ad in ads:
                ad.extract()
            
            # 장소 이름 추출 및 출력
            titles = soup.select("a.link_name")
            for title in titles:
                print(title.text)
        
        # 다음 버튼 클릭
        next_button = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.page\\.next")
        if next_button.get_attribute("class") == "disabled":
            break
        next_button.click()
        time.sleep(2)
except Exception as e:
    print(f"오류 발생: {e}")
    print("수집이 완료되었습니다!")

# 드라이버 닫기
driver.quit()
```

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup as bs
import pandas as pd

# webdriver_manager를 사용하여 Chrome 드라이버 설정
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# 카카오 지도 웹사이트 열기
driver.get("https://map.kakao.com/")

# 요소가 나타날 때까지 기다린 후 클릭
time.sleep(2)
img = driver.find_element(By.CSS_SELECTOR, "div.inner_coach_layer")
img.click()

# 검색 수행
search = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.query")
search.send_keys("쏘카존 강원특별자치도")
search_btn = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.submit")
search_btn.click()

# 검색 결과가 로드될 시간을 줌
time.sleep(2)

# BeautifulSoup을 사용하여 페이지 소스 파싱
soup = bs(driver.page_source, "lxml")

# 광고 제거
ads = soup.select("span.ico_ad")
for ad in ads:
    ad.extract()

# "장소 더보기" 버튼 클릭
btn_more = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.place\\.more")
btn_more.click()

# 새로운 결과가 로드될 시간을 줌
time.sleep(2)

# 결과를 저장할 리스트 생성
data = []

try:
    while True:
        for page_num in range(1, 6):
            # 페이지 버튼 클릭 후 데이터 스크랩
            page_button = driver.find_element(By.CSS_SELECTOR, f"#info\\.search\\.page\\.no{page_num}")
            page_button.click()
            time.sleep(2)
            
            soup = bs(driver.page_source, "lxml")
            
            # 광고 제거
            ads = soup.select("span.ico_ad")
            for ad in ads:
                ad.extract()
            
            # 장소 이름과 도로명 주소 추출
            titles = soup.select("a.link_name")
            addresses = soup.select("p[data-id='address']")
            
            for title, address in zip(titles, addresses):
                data.append({
                    "장소": title['title'],
                    "도로명 주소": address['title']
                })
        
        # 다음 버튼 클릭
        next_button = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.page\\.next")
        if next_button.get_attribute("class") == "disabled":
            break
        next_button.click()
        time.sleep(2)
except Exception as e:
    print(f"오류 발생: {e}")
    print("수집이 완료되었습니다!")

# 드라이버 닫기
driver.quit()

# 데이터를 DataFrame으로 변환
df = pd.DataFrame(data)

# DataFrame을 CSV 파일로 저장
df.to_csv("socar_locations강원특별자치도.csv", index=False, encoding='utf-8-sig')

print("데이터가 'socar_locations.csv' 파일에 저장되었습니다.")
```

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup as bs
import pandas as pd

# webdriver_manager를 사용하여 Chrome 드라이버 설정
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

# 검색할 구 목록
districts = [
    "연천", "포천", "파주", "동두천", "양주", "의정부", "가평", "고양", "구리", "남양주", 
    "하남", "양평", "광주", "여주", "김포", "부천", "광명", "시흥", "안양", "과천", 
    "의왕", "군포", "수원", "안산", "성남", "용인", "이천", "화성", "오산", "평택", 
    "안성"
]


# 결과를 저장할 리스트 생성
data = []

for district in districts:
    # 카카오 지도 웹사이트 열기
    driver.get("https://map.kakao.com/")
    
    # 요소가 나타날 때까지 기다린 후 클릭
    time.sleep(2)
    try:
        img = driver.find_element(By.CSS_SELECTOR, "div.inner_coach_layer")
        img.click()
    except:
        pass
    
    # 검색 수행
    search = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.query")
    search.send_keys(f"쏘카존 경기도 {district}")
    search_btn = driver.find_element(By.CSS_SELECTOR, "#search\\.keyword\\.submit")
    search_btn.click()
    
    # 검색 결과가 로드될 시간을 줌
    time.sleep(2)
    
    # BeautifulSoup을 사용하여 페이지 소스 파싱
    soup = bs(driver.page_source, "lxml")
    
    # 광고 제거
    ads = soup.select("span.ico_ad")
    for ad in ads:
        ad.extract()
    
    # "장소 더보기" 버튼 클릭
    try:
        btn_more = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.place\\.more")
        btn_more.click()
        time.sleep(2)
    except:
        pass
    
    try:
        while True:
            # 1부터 5까지 페이지를 반복
            for page_num in range(1, 6):
                try:
                    page_button = driver.find_element(By.CSS_SELECTOR, f"#info\\.search\\.page\\.no{page_num}")
                    page_button.click()
                    time.sleep(2)
                    
                    soup = bs(driver.page_source, "lxml")
                    
                    # 광고 제거
                    ads = soup.select("span.ico_ad")
                    for ad in ads:
                        ad.extract()
                    
                    # 장소 이름과 도로명 주소 추출
                    titles = soup.select("a.link_name")
                    addresses = soup.select("p[data-id='address']")
                    
                    for title, address in zip(titles, addresses):
                        data.append({
                            "장소": title['title'],
                            "도로명 주소": address['title'],
                            "지역": district
                        })
                except:
                    break
            
            # 다음 버튼 클릭
            next_button = driver.find_element(By.CSS_SELECTOR, "#info\\.search\\.page\\.next")
            if "disabled" in next_button.get_attribute("class"):
                break
            next_button.click()
            time.sleep(2)
    except Exception as e:
        print(f"오류 발생: {e}")

# 드라이버 닫기
driver.quit()

# 데이터를 DataFrame으로 변환
df = pd.DataFrame(data)

# DataFrame을 CSV 파일로 저장
df.to_csv("socar_locations_geongi.csv", index=False, encoding='utf-8-sig')

print("데이터가 'socar_locations_geongi.csv' 파일에 저장되었습니다.")
```

```python
import pandas as pd

# 두 개의 CSV 파일 읽기
df1 = pd.read_csv('socar_locations_combined_unique_final.csv', encoding='utf-8-sig')
df2 = pd.read_csv('socar_locations_geo_all.csv', encoding='utf-8-sig')

# 두 데이터프레임 합치기
df_combined = pd.concat([df1, df2])

# 열 이름을 적절히 수정
df_combined.columns = ['장소', '도로명 주소', '위도', '경도']

# 1. 열 1('장소')에서 중복 제거 (위도와 경도가 존재하는 행을 우선으로 남김)
df_combined = df_combined.sort_values(by=['위도', '경도'], na_position='last')
df_combined = df_combined.drop_duplicates(subset=['장소'], keep='first')

# 2. 열 2('도로명 주소')에서 중복 제거 (위도와 경도가 존재하는 행을 우선으로 남김)
df_combined = df_combined.sort_values(by=['위도', '경도'], na_position='last')
df_combined = df_combined.drop_duplicates(subset=['도로명 주소'], keep='first')

# 결과 저장
df_combined.to_csv('socar_locations_cleaned_last.csv', index=False, encoding='utf-8-sig')

print("중복 제거 및 정리 완료. 결과는 'socar_locations_cleaned_last.csv'에 저장되었습니다.")
```
