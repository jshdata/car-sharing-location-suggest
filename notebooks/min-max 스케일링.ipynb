```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

file_path2 = "C:/Users/kfq/Desktop/P-MEDIAN 알고리즘/combination_results_2.csv"
df2 = pd.read_csv(file_path2)
df2["Weighted Distance Sum"] = df2[["접근성", "수요성", "통행량", "가용성", "활동성"]].sum(axis=1)

columns_to_scale = ['접근성', '수요성', '통행량', '가용성', '활동성']
df2_scaled = df2.copy()
scaler = MinMaxScaler()

df2_scaled[columns_to_scale] = df2_scaled[columns_to_scale].round(2)

def assign_scores(column):
    return pd.qcut(column, q=5, labels=[5, 4, 3, 2, 1])

# 각 지표에 대해 점수 부여
for col in columns_to_scale:
    df2_scaled[col] = assign_scores(df2_scaled[col])
df2_sorted = df2_scaled.sort_values(by="Weighted Distance Sum")

# 총 10개의 행을 추출
num_rows = len(df2_sorted)
selected_rows = []
for i in range(10):
    start_idx = int(i * 0.1 * num_rows)
    selected_rows.append(df2_sorted.iloc[start_idx])

df2_selected = pd.DataFrame(selected_rows)
print(df2_selected)
```
