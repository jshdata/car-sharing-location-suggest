```python
# CSV 파일 불러오기
pop_data = pd.read_csv("C:/Users/kfq/Downloads/대전광역시 구별 유동인구 수(2020.12).csv")
# 시군구 열 이름을 일치시킵니다.
pop_data.rename(columns={'시군구': 'SGG_NM'}, inplace=True)

# 20대, 30대 컬럼 이름 변경
pop_data.rename(columns={'20대': '20_year', '30대': '30_year'}, inplace=True)

# GeoDataFrame에 인구 데이터 병합
cty_shp = cty_shp.merge(pop_data[['SGG_NM', '20_year', '30_year']], on='SGG_NM', how='left')
cty_shp = cty_shp.to_crs(epsg=4326)
# 결과를 GeoJSON 파일로 저장
cty_shp.to_file("C:/Users/kfq/Downloads/대전광역시_유동인구_2020_12.geojson", driver='GeoJSON')
```
