```python
import pandas as pd

# 예시 데이터프레임 (생략)
import pandas as pd
# 숫자로 변환 가능한 컬럼들을 모두 숫자로 변환
cols_to_convert = ['승용차추정교통량', '주차구획수', '20대인구_격자', '30대인구_격자', '숙소별 관광지 거리 평균']

# ','를 제거하고 숫자로 변환, 오류 발생시 NaN으로 처리
for col in cols_to_convert:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', ''), errors='coerce')

# 숫자형 컬럼들만 선택
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

# 역순 범주화를 적용할 컬럼
reverse_cols = ['시내버스정류장', '도시철도', '차고지']

# 각 컬럼에 대해 범주화
for col in numeric_cols:
    # 0 값을 제외하고 범주화하기 위해 0을 NaN으로 변환
    df[col] = df[col].replace(0, pd.NA)
    
    # 역순으로 범주화할 컬럼과 일반 범주화를 구분
    if col in reverse_cols:
        # 역순 범주화 (작은 값이 더 좋은 값일 경우)
        df[col] = pd.cut(df[col].dropna(), bins=5, labels=[5, 4, 3, 2, 1]).astype(pd.Int64Dtype())
    else:
        # 일반 범주화 (큰 값이 더 좋은 값일 경우)
        df[col] = pd.cut(df[col].dropna(), bins=5, labels=[1, 2, 3, 4, 5]).astype(pd.Int64Dtype())

# 결과 출력
print(df)
```

```python

# 접근성 지표 계산
df['접근성'] = df[['시내버스정류장', '도시철도', '시외버스터미널', '기차역', '숙소']].mean(axis=1, skipna=True)

# 수요성 지표 계산
df['수요성'] = df[['20대인구_격자', '30대인구_격자', '숙소별 관광지 거리 평균']].mean(axis=1, skipna=True)

# 통행량 지표 계산
df['통행량'] = df[['승용차추정교통량']].mean(axis=1, skipna=True)

# 가용성 지표 계산
df['가용성'] = df[['공영주차장', '차고지', '주차구획수']].mean(axis=1, skipna=True)

# 활동성 지표 계산
df['활동성'] = df[['20대 구별 유동인구', '30대 구별 유동인구', '상권매출액']].mean(axis=1, skipna=True)

# 결과 확인
print(df[['접근성', '수요성', '통행량', '가용성', '활동성']].head())
```
