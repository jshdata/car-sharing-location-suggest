```python
import geopandas as gpd

# GeoJSON 파일 경로
file_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\지도\대전광역시 행정동.geojson"

# GeoJSON 파일을 읽어옵니다.
gdf = gpd.read_file(file_path)
gdf['hgd_nm'] = gdf['adm_nm'].apply(lambda x: x.split()[2])

# 데이터프레임 정보 출력
print(gdf)

# 데이터 시각화 (필요한 경우)
gdf.plot()
```

```python
# CSV 파일 경로
csv_file_path = r"C:\Users\kfq\Downloads\대전_구별_동별_연령별_인구 수.csv"
# CSV 파일을 읽어옵니다.
df = pd.read_csv(csv_file_path)
# "adm_nm" 열을 시, 구, 동으로 분리하고 세 번째 값(동)을 "hgd_nm"이라는 새로운 열에 저장
gdf['hgd_nm'] = gdf['adm_nm'].apply(lambda x: x.split()[2])

# CSV 데이터의 컬럼 이름을 변경합니다.
df.rename(columns={'20대': '20_year', '30대': '30_year', '총인구': 'all_year'}, inplace=True)

# GeoDataFrame과 DataFrame을 "행정동"과 "hgd_nm"을 기준으로 병합합니다.
merged_gdf = gdf.merge(df[['행정동', '20_year', '30_year', "all_year"]], how='left', left_on='hgd_nm', right_on='행정동')

# 병합된 데이터프레임 확인
print(merged_gdf.head())
```

```python
# GeoJSON 파일로 저장할 경로 설정
output_geojson_path = r"C:\Users\kfq\Downloads\merged_data.geojson"

# GeoDataFrame을 GeoJSON 파일로 저장
merged_gdf.to_file(output_geojson_path, driver='GeoJSON')
```
