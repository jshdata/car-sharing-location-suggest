```python
import pandas as pd
import os

# 파일 경로 설정
file_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\관광지\대전광역시 관광지 구별 시각화.csv"

# 파일 존재 여부 확인
if os.path.exists(file_path):
    try:
        # CSV 파일을 CP949로 불러오기
        df = pd.read_csv(file_path, encoding='cp949')
        
        # 같은 경로에 UTF-8-SIG로 저장하기
        df.to_csv(file_path, encoding='utf-8-sig', index=False)
        print("파일이 성공적으로 저장되었습니다.")
    except UnicodeDecodeError:
        print("인코딩 오류 발생. 파일 인코딩을 확인해주세요.")
else:
    print("지정된 경로에 파일이 없습니다. 경로를 다시 확인해 주세요.")
```
