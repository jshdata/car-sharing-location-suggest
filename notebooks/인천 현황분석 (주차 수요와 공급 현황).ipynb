```python
# 자동차 등록 대수, 공영 주차장
# 구별 공영주차장 비율
# 구별 등록대수 비율
# 등록대수가 적고, 공영주차장 비율이 많은 곳
# 공영주차장 비율 / 등록대수 비율 이 큰 구
import pandas as pd
car = pd.read_csv("C:/Users/kfq/Downloads/1ST_Final_Project/카셰어링/데이터/인천 데이터/전국 자동차 등록 대수 현황 (2024).csv")
parking = pd.read_csv("C:/Users/kfq/Downloads/인천광역시 공영주차장 데이터.csv", encoding = "euc-kr")
```

```python
import matplotlib.pyplot as plt
import matplotlib
import koreanize_matplotlib
import warnings

# 경고 무시 설정
warnings.filterwarnings("ignore")

# 한글 폰트 설정
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 윈도우에서는 'Malgun Gothic'을 사용합니다.
# matplotlib.rcParams['font.family'] = 'AppleGothic'  # macOS에서는 'AppleGothic'을 사용합니다.
# matplotlib.rcParams['font.family'] = 'NanumGothic'  # Ubuntu와 같은 리눅스에서는 'NanumGothic'을 사용합니다.
matplotlib.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지
```

```python
parking['시군구'] = parking['제공기관명'].apply(lambda x: x.split()[1] if len(x.split()) > 1 else None)

parking
```

```python
print(car)
```

```python
car['총계'] = car['총계'].str.replace(',', '').astype(int)

# 시군구별 차량등록비율 컬럼 추가 (백분율로 변환)
total_sum = car['총계'].sum()
car['시군구별 차량등록비율'] = round((car['총계'] / total_sum) * 100, 2)

# 결과 확인
print(car)
```

```python
grouped_parking = parking.groupby('시군구')['주차구획수'].sum().reset_index()

print(grouped_parking)
```

```python
# 시군구별 주차구획 비율 컬럼 추가 (백분율로 변환)
total_spaces = grouped_parking['주차구획수'].sum()
grouped_parking['주차구획 비율'] = (grouped_parking['주차구획수'] / total_spaces) * 100

# 결과 확인
print(grouped_parking)
```

```python
# 시군구를 기준으로 두 데이터프레임 결합
merged_data = pd.merge(car, grouped_parking, on='시군구')

# 결과 확인
print(merged_data)
```

```python
# "비율 비교" 컬럼 추가 (주차구획 비율 / 차량등록비율)
merged_data['비율 비교'] = merged_data['주차구획 비율'] / merged_data['시군구별 차량등록비율']

# 결과 확인
print(merged_data)
```

```python
# 비율 비교를 기준으로 내림차순 정렬
merged_data = merged_data.sort_values(by='비율 비교', ascending=False)

# 수직 막대그래프 생성
plt.figure(figsize=(10, 6))
plt.bar(merged_data['시군구'], merged_data['비율 비교'], color='skyblue')
plt.ylabel('비율 비교 (주차구획 비율 / 차량등록비율)')
plt.xlabel('시군구')
plt.title('시군구별 주차구획 비율 대비 차량등록 비율 비교')
plt.xticks(rotation=45, ha='right')  # x축 레이블 회전
plt.tight_layout()  # 레이아웃 자동 조정
plt.show()
```

```python
# 비율 비교를 기준으로 내림차순 정렬
merged_data = merged_data.sort_values(by='비율 비교', ascending=False)

# 평균 비율 비교 계산
mean_ratio = merged_data['비율 비교'].mean()

# 수직 막대그래프 생성
plt.figure(figsize=(12, 8))
bars = plt.bar(merged_data['시군구'], merged_data['비율 비교'], color='skyblue')

# 평균선 추가
plt.axhline(mean_ratio, color='red', linestyle='--', linewidth=1.5)

# 평균선 텍스트 추가
plt.annotate(
    f'평균: {mean_ratio:.2f}', 
    xy=(0.95, mean_ratio), 
    xytext=(0.95, mean_ratio + 1),
    textcoords='offset points',
    ha='right', 
    va='bottom',
    color='red',
    fontsize=12,
    bbox=dict(facecolor='white', edgecolor='red', boxstyle='round,pad=0.5')
)

# 축 레이블 및 제목 설정
plt.ylabel('비율 비교 (주차구획 비율 / 차량등록비율)')
plt.xlabel('시군구')
plt.title('시군구별 주차구획 비율 대비 차량등록 비율 비교')
plt.xticks(rotation=45, ha='right')  # x축 레이블 회전
plt.tight_layout()  # 레이아웃 자동 조정

# 그래프 출력
plt.show()
```
