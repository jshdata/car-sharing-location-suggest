```python
# 상위 10개
import pandas as pd

# 파일 경로를 변수로 지정
path_combination_results_1 = r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\combination_results_1.csv"
path_combination_results_2 = r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\combination_results_2.csv"
path_combination_results_3 = r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\combination_results_3.csv"
path_grid_generation = r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\격자생성.csv"

# CSV 파일을 DataFrame으로 읽어오기
df1 = pd.read_csv(path_combination_results_1)
df2 = pd.read_csv(path_combination_results_2)
df3 = pd.read_csv(path_combination_results_3)
df = pd.read_csv(path_grid_generation)

# 'Weighted Distance Sum' 열 추가
df1["Weighted Distance Sum"] = df1[["접근성", "수요성", "통행량", "가용성", "활동성"]].sum(axis=1)
df2["Weighted Distance Sum"] = df2[["접근성", "수요성", "통행량", "가용성", "활동성"]].sum(axis=1)
df3["Weighted Distance Sum"] = df3[["접근성", "수요성", "통행량", "가용성", "활동성"]].sum(axis=1)

# 상위 10개 행 추출
df1_sorted_top10 = df1.sort_values(by='Weighted Distance Sum').head(10)
df2_sorted_top10 = df2.sort_values(by='Weighted Distance Sum').head(10)
df3_sorted_top10 = df3.sort_values(by='Weighted Distance Sum').head(10)

# Combination 열 정리
def clean_combination(combination):
    combination = combination.strip('()').replace(' ', '')  # 괄호와 공백 제거
    if combination.endswith(','):
        combination = combination[:-1]  # 끝의 쉼표 제거
    return combination

df1_sorted_top10['Combination'] = df1_sorted_top10['Combination'].apply(clean_combination)
df2_sorted_top10['Combination'] = df2_sorted_top10['Combination'].apply(clean_combination)
df3_sorted_top10['Combination'] = df3_sorted_top10['Combination'].apply(clean_combination)

# 조합 정보 추출 함수
def extract_combination_info(df, combination_df):
    result_dfs = []
    
    for _, row in combination_df.iterrows():
        combination = row['Combination']
        indices = [int(i) for i in combination.split(',') if i.isdigit()]  # 정수로 변환할 수 있는 값만 포함

        combined_data = {}
        for i, idx in enumerate(indices):
            if idx < len(df):  # 인덱스 범위 체크
                row_data = df.iloc[idx]
                for col in df.columns:
                    combined_data[f"{col}{i+1}"] = row_data[col]
            else:
                print(f"Index {idx} is out of range for the DataFrame.")
        
        selected_df = pd.DataFrame([combined_data])
        result_dfs.append(selected_df)
    
    final_result = pd.concat(result_dfs, ignore_index=True)
    return final_result

# 조합 정보 추출
df1_result = extract_combination_info(df, df1_sorted_top10)
df2_result = extract_combination_info(df, df2_sorted_top10)
df3_result = extract_combination_info(df, df3_sorted_top10)

# 'Weighted Distance Sum' 열 추가
def add_weighted_distance_sum(result_df, combination_df):
    weighted_sums = combination_df['Weighted Distance Sum'].values
    result_df['Weighted Distance Sum'] = weighted_sums
    return result_df

df1_result = add_weighted_distance_sum(df1_result, df1_sorted_top10)
df2_result = add_weighted_distance_sum(df2_result, df2_sorted_top10)
df3_result = add_weighted_distance_sum(df3_result, df3_sorted_top10)

# 결과를 CSV 파일로 저장
df1_result.to_csv(r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\df1_combination_info_top10.csv", index=False, encoding="utf-8-sig")
df2_result.to_csv(r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\df2_combination_info_top10.csv", index=False, encoding="utf-8-sig")
df3_result.to_csv(r"C:\Users\kfq\Desktop\P-MEDIAN 알고리즘\df3_combination_info_top10.csv", index=False, encoding="utf-8-sig")

# 결과 출력
print(df1_result.head())
print(df2_result.head())
# print(df3_result.head())
```
