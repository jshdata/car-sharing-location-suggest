```python
import pandas as pd

# 파일 경로 설정
car_sharing_path = r"C:/Users/kfq/Downloads/1ST_Final_Project/카셰어링/데이터/대전 데이터/관광지/대전광역시 관광지 데이터.csv"
train_station_path = r"C:/Users/kfq/Downloads/1ST_Final_Project/카셰어링/데이터/대전 데이터/상권/대전광역시 상가 정보 데이터/대전광역시 숙소 정보 데이터.csv"

# 데이터 불러오기
df1 = pd.read_csv(car_sharing_path, encoding = "euc-kr")
df2 = pd.read_csv(train_station_path, encoding = "cp949")

# Cartesian Product 수행
df1['key'] = 1
df2['key'] = 1

df_combined = pd.merge(df1, df2, on='key').drop('key', axis=1)

# 결과 확인
print("결과 데이터프레임:")
print(df_combined.head())
print(f"총 행 수: {len(df_combined)}")

output_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\대중교통\df_combined_관광지_숙소.csv"
df_combined.to_csv(output_path, index=False, encoding='utf-8-sig')
```
