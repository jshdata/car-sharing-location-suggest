```python
import pandas as pd
import xml.etree.ElementTree as ET
from urllib.request import Request, urlopen
from urllib.parse import urlencode, quote_plus

# 기본 URL (공백 제거 및 인코딩)
url = "https://api.vworld.kr/ned/data/getIndvdLandPrice"

# 요청 변수 설정
params = {
    "format": "xml",
    "key": "5970838A-B5F0-3594-8DED-50DA74A34CD2",
    "domain": "도메인",
    "stdrYear": "2022",
    "reqLvl": "3",
    "ldCode": "4793025021",
    "numOfRows": "1000",
    "pageNo": "1"
}

# URL 인코딩
queryParams = "?" + urlencode(params, quote_via=quote_plus)

# 전체 URL 생성
full_url = url + queryParams

# 요청 및 응답
request = Request(full_url)
request.get_method = lambda: "GET"
response_body = urlopen(request).read()

# XML 파싱
root = ET.fromstring(response_body)

# 데이터 추출
fields = root.find('.//fields')
data = []

for field in fields.findall('field'):
    row = {child.tag: child.text for child in field}
    data.append(row)

# DataFrame 생성
df = pd.DataFrame(data)

# DataFrame 출력
print(df)
```
