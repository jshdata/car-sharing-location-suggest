```python
import pandas as pd
df = pd.read_csv("C:/Users/kfq/Downloads/1ST_Final_Project/카셰어링/데이터/인천광역시_도로 주변 유동인구_20211130/인천광역시_유동인구_202111.csv", encoding = "euc-kr")
```

```python
df["동"].value_counts()
```

```python
value_counts_df = df["동"].value_counts().to_frame().reset_index()
value_counts_df.columns = ["동", "도로주변 유동인구수"]

pd.set_option('display.max_rows', None)

print(value_counts_df)
```

```python
value_counts_df2 = df["군구"].value_counts().to_frame().reset_index()
value_counts_df2.columns = ["구", "도로주변 유동인구수"]

pd.set_option('display.max_rows', None)

print(value_counts_df2)
```

```python
value_counts_df.to_csv('인천광역시 동별 도로주변 유동인구수.csv', index=False, encoding = "euc-kr")
value_counts_df2.to_csv('인천광역시 구별 도로주변 유동인구수.csv', index=False, encoding = "euc-kr")
```
