```python
import pandas as pd
import requests

# Google Geocoding API 키
API_KEY = 'YOUR_CREDENTIAL_HERE'
BASE_URL = 'https://maps.googleapis.com/maps/api/geocode/json'

def get_lat_lon(address):
    """주소를 입력 받아 위도와 경도를 반환하는 함수"""
    url = f"{BASE_URL}?address={address}&key={API_KEY}"
    response = requests.get(url).json()
    if response['status'] == 'OK':
        location = response['results'][0]['geometry']['location']
        return location['lat'], location['lng']
    else:
        return None, None

# 데이터 입력
data = df_melted

# DataFrame 생성
df = df_melted

# 승차역과 도착역의 위도와 경도를 구할 주소 목록
stations = list(set(data['승차역'] + data['도착역']))

# 주소에서 위도와 경도 추출
locations = {}
for station in stations:
    lat, lon = get_lat_lon(station + '역')  # '역'을 주소에 추가하여 정확한 위치를 찾도록 시도
    if lat and lon:
        locations[station] = (lat, lon)
    else:
        locations[station] = (None, None)

# 결과를 DataFrame으로 변환
df_locations = pd.DataFrame.from_dict(locations, orient='index', columns=['위도', '경도']).reset_index()
df_locations.rename(columns={'index': '역명'}, inplace=True)

# 시군구와 행정동 정보 추출을 위한 Geocoding 함수
def get_location_info(lat, lon):
    """위도와 경도를 입력 받아 시군구와 행정동 정보를 반환하는 함수"""
    url = f"{BASE_URL}?latlng={lat},{lon}&key={API_KEY}"
    response = requests.get(url).json()
    if response['status'] == 'OK':
        address = response['results'][0]['address_components']
        city = next((comp['long_name'] for comp in address if 'administrative_area_level_2' in comp['types']), None)
        district = next((comp['long_name'] for comp in address if 'administrative_area_level_3' in comp['types']), None)
        return city, district
    else:
        return None, None

# 위도와 경도에 대한 시군구와 행정동 정보 추출
df_locations[['시군구', '행정동']] = df_locations.apply(lambda row: get_location_info(row['위도'], row['경도']), axis=1, result_type='expand')

# 원본 데이터와 시군구, 행정동 정보를 병합
df_final = pd.merge(df, df_locations, left_on='승차역', right_on='역명', how='left')

# 결과를 CSV 파일로 저장
df_final.to_csv('대전역_승차_정보.csv', index=False, encoding='utf-8-sig')

print("CSV 파일이 생성되었습니다.")
```

```python
from geopy.geocoders import Nominatim
import pandas as pd

# Nominatim 설정
geolocator = Nominatim(user_agent="geoapiExercises")

# CSV 파일 불러오기
file_path = "C:/Users/kfq/Downloads/대전광역시 상권매출.csv"
df = pd.read_csv(file_path)

# 컬럼명 수정 (필요시)
df.columns = ['주소지', '매출액', '주소', '매출액_숫자', '위도', '경도', '시군구']

# 행정동 추출 함수 정의
def get_administrative_dong(lat, lng):
    location = geolocator.reverse((lat, lng), language='ko')
    if location and 'suburb' in location.raw['address']:
        return location.raw['address']['suburb']
    elif location and 'neighbourhood' in location.raw['address']:
        return location.raw['address']['neighbourhood']
    return None

# 데이터프레임에 행정동 추가
df['행정동'] = df.apply(lambda row: get_administrative_dong(row['위도'], row['경도']), axis=1)

# 결과 출력
print(df.head())
```

```python
import pandas as pd

# CSV 파일 불러오기
file_path = "C:/Users/kfq/Downloads/대전광역시 상권매출.csv"
df = pd.read_csv(file_path)

# 주소 컬럼을 분리하여 새로운 컬럼으로 추가
def split_address(address):

    parts = address.split()
    if len(parts) >= 3:
        시도 = parts[0]
        시군구 = parts[1]
        도로명 = parts[2]
        return pd.Series([시도, 시군구, 도로명])
    else:
        # 주소 정보가 부족할 경우 빈 문자열 반환
        return pd.Series([None, None, None])

# 새로운 컬럼 추가
df[['시도', '시군구', '도로명']] = df['주소'].apply(split_address)


output_file_path = "C:/Users/kfq/Downloads/대전광역시 상권매출_분리주소.csv"
df.to_csv(output_file_path, index=False, encoding='utf-8-sig')

print(f"파일이 성공적으로 저장되었습니다: {output_file_path}")
```
