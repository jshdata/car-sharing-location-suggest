```python
import pandas as pd

# CSV 파일 경로
file_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\대중교통\대전광역시 도시철도 시간대별 이동량.csv"

# CSV 파일 불러오기
data = pd.read_csv(file_path)

# "시간 숫자열"과 "시간 문자열"을 시간만으로 변환
data['시간 timestamp'] = pd.to_timedelta(data['시간 숫자열'], unit='h')
data['시간 timestamp'] = pd.to_datetime('2024-01-01') + data['시간 timestamp']

data['시간 숫자열 타임스탬프'] = pd.to_datetime('2024-01-01') + pd.to_timedelta(data['시간 숫자열'], unit='h')
data['시간 숫자열 타임스탬프'] = pd.to_datetime(data['시간 숫자열 타임스탬프'])

# 변환된 데이터를 CSV 파일로 저장
output_file_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\대중교통\대전광역시 도시철도 시간대별 이동량_변환2.csv"
data.to_csv(output_file_path, index=False, encoding='utf-8-sig')

print(f"CSV 파일이 저장되었습니다: {output_file_path}")
```
