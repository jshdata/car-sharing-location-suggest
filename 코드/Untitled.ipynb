```python
import pandas as pd

# CSV 파일 경로
input_file = r"C:\Users\kfq\Downloads\대전광역시 구별 유동인구 수(2020.12) (1).csv"

# CSV 파일 불러오기
df = pd.read_csv(input_file, encoding='utf-8')  # 인코딩이 필요하면 'utf-8' 또는 'cp949'로 조정

# 기준일자와 시군구명으로 그룹화하고 보정보행인구수와 보정비보행인구수의 합을 계산
grouped_df = df.groupby(['기준일자', '시군구명']).agg({
    '보정보행인구수': 'sum',
    '보정비보행인구수': 'sum'
}).reset_index()

# 새로운 CSV 파일로 저장
output_file = r"C:\Users\kfq\Desktop\대전광역시_유동인구_요약.csv"
grouped_df.to_csv(output_file, index=False, encoding='utf-8-sig')  # 인코딩 필요시 'utf-8' 또는 'cp949'로 조정

print(f"CSV 파일이 {output_file}에 저장되었습니다.")
```

```python
import pandas as pd

# 기존 요약된 CSV 파일 경로
input_file = r"C:\Users\kfq\Desktop\대전광역시_유동인구_요약.csv"

# CSV 파일 불러오기
df = pd.read_csv(input_file, encoding='utf-8')

# Melt 형식으로 데이터 변환
melted_df = df.melt(id_vars=['시군구', '기준일자'], 
                    value_vars=['보정보행인구수', '보정비보행인구수'], 
                    var_name='인구수 종류', 
                    value_name='인구수')

# 피벗 테이블 생성
pivot_df = melted_df.pivot(index=['시군구', '인구수 종류'], 
                           columns='기준일자', 
                           values='인구수')

# 새로운 CSV 파일로 저장
output_file = r"C:\Users\kfq\Desktop\대전광역시_유동인구_피벗_최종.csv"
pivot_df.to_csv(output_file, encoding='utf-8-sig')

print(f"CSV 파일이 {output_file}에 저장되었습니다.")
```

```python
import pandas as pd

# 파일 경로
file_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\상권\거리데이터 숙소Ｘ관광지\결과\관광지X숙소 거리데이터.csv"

# 데이터 불러오기
df = pd.read_csv(file_path)

# '관광지와 숙소의 거리' 컬럼을 오름차순 정렬
df_sorted = df.sort_values(by=["관광지와 숙소의 거리"], ascending=True)

# 인덱스 초기화
df_sorted.reset_index(drop=True, inplace=True)

# 거리 기준으로 사분위수 계산
quantiles = df_sorted["관광지와 숙소의 거리"].quantile([0.25, 0.5, 0.75])

# 각 사분위에 따라 데이터프레임 분할
df_closest = df_sorted[df_sorted["관광지와 숙소의 거리"] <= quantiles[0.25]]
df_close = df_sorted[(df_sorted["관광지와 숙소의 거리"] > quantiles[0.25]) & (df_sorted["관광지와 숙소의 거리"] <= quantiles[0.5])]
df_far = df_sorted[(df_sorted["관광지와 숙소의 거리"] > quantiles[0.5]) & (df_sorted["관광지와 숙소의 거리"] <= quantiles[0.75])]
df_farthest = df_sorted[df_sorted["관광지와 숙소의 거리"] > quantiles[0.75]]

# 각 데이터프레임을 지정된 경로에 CSV 파일로 저장
output_path = r"C:\Users\kfq\Downloads\1ST_Final_Project\카셰어링\데이터\대전 데이터\상권\거리데이터 숙소Ｘ관광지\결과"

df_closest.to_csv(f"{output_path}\\가까움.csv", index=False, encoding='utf-8-sig')
df_close.to_csv(f"{output_path}\\조금 가까움.csv", index=False, encoding='utf-8-sig')
df_far.to_csv(f"{output_path}\\조금 멈.csv", index=False, encoding='utf-8-sig')
df_farthest.to_csv(f"{output_path}\\멈.csv", index=False, encoding='utf-8-sig')
```
