```python
# 주소 컬럼을 분리하여 새로운 컬럼으로 추가
def split_address(address):

    parts = address.split()
    if len(parts) >= 3:
        시도 = parts[0]
        시군구 = parts[1]
        행정동 = parts[2]
        return pd.Series([시도, 시군구, 행정동])
    else:
        # 주소 정보가 부족할 경우 빈 문자열 반환
        return pd.Series([None, None, None])

# 새로운 컬럼 추가
df[['시도', '시군구', '행정동']] = df['행정기관'].apply(split_address)
df

output_file_path = "C:/Users/kfq/Downloads/분리주소.csv"
df.to_csv(output_file_path, index=False, encoding='utf-8-sig')
```
